```python
import random
import numpy as np
```

# Envirenment

```python
NUM_STATES = 5                           # Number of states
NUM_ACTIONS = 2

ALPHA = 0.1
GAMMA = 0.9
EPSILON = 0.1

START_STATE = 0
GOAL_STATE = 4
TERMINAL_STATE = 4                          

LEFT = 0
RIGHT = 1
```

```python
def step(state, action, goal_state=GOAL_STATE):
    """
    Applies an action and returns: next_state, reward, terminated
    """
    if action == 0:
        state -= 1
        if state < 0 : state = 0
    elif action == 1:
        state += 1
        if state > goal_state: state = goal_state
    
    if state == goal_state:
        reward = 1
    else:
        reward = 0
    
    if state == goal_state: terminated = True
    else: terminated = False
    
    return state, reward, terminated
```

# Policy

```python
def greedy_action(state, Q):
    """
    Return the action with the larger estimated Q-value for specified state.
    """
    if Q[state, 0] < Q[state, 1]:
        return int(1)
    elif Q[state, 0] > Q[state, 1]:
        return int(0)
    else:
        return int(random.randint(0, 1))
```

```python
def choose_action(state, Q, epsilon=EPSILON):
    """
    Choose an action using epsilon-greedy exploration.
    """
    if random.random() < epsilon:
        return int(random.randint(0, 1))
    else:
        return greedy_action(state, Q)
```

# Q-Learning

```python
def q_learning_update(
        Q, state, action, reward, 
        next_state, done, alpha=ALPHA, gamma=GAMMA
):
    """
    Q-Learning updating algorithm
    """
    if done:
        target = reward
    else:
        best_next_value = np.max([Q[next_state, 0], Q[next_state, 1]])
        target = reward + gamma * best_next_value
    
    td_error = target - Q[state, action]
    Q[state, action] += alpha * td_error

    return Q, td_error
```

# Episode Generation

```python
def run_episode(Q, epsilon=EPSILON, alpha=ALPHA, gamma=GAMMA ,max_steps=100):
    """
    Runs one episode and return its total rewards, number of steps until termination, Q matrix, and td error for every step.
    """
    state = START_STATE
    total_reward = 0
    td_error = []                                                       

    for step_idx in range(max_steps):
        action = choose_action(state, Q, epsilon=epsilon)
        next_state, reward, terminated = step(state, action, GOAL_STATE)
        Q, error = q_learning_update(Q=Q,
                                        state=state,
                                        action=action,
                                        reward=reward,
                                        next_state=next_state,
                                        done=terminated,
                                        alpha=alpha, gamma=gamma,)
        td_error.append(error)
        total_reward += reward
        state = next_state
        if terminated:
            return total_reward, step_idx + 1, Q, td_error    
    
    return total_reward, max_steps, Q, td_error  
```

# Training

```python
Q = np.zeros((NUM_STATES, NUM_ACTIONS), dtype=float)
num_episodes = 5000
episode_length = []
rewards = []
errors = []
for episode in range(num_episodes):
    reward, length, Q, td_error = run_episode(Q, EPSILON, ALPHA, GAMMA)
    episode_length.append(length)
    rewards.append(reward)
    errors.append(td_error)
```

# Tests

```python
# Environment
assert step(0, RIGHT) == (1, 0, False)
assert step(3, RIGHT) == (4, 1, True)

# Learned ordering
for state in range(4):
    assert Q[state, RIGHT] > Q[state, LEFT]

# Goal transition
assert np.isclose(Q[3, RIGHT], 1.0)
```

# Evaluation

```python
def evaluation(Q, num_episodes=100, max_steps=100):
   
    episode_length = []
    episode_rewards = []
  
    for episode in range(num_episodes):

        state = START_STATE
        length = max_steps    
        total_reward = 0                                             

        for step_idx in range(max_steps):
            action = greedy_action(state, Q)
            next_state, reward, terminated = step(state, action, GOAL_STATE)
            
            total_reward += reward
            state = next_state
            if terminated:
                length = step_idx + 1
                break  
    
        episode_length.append(length)
        episode_rewards.append(total_reward)
    return episode_length, episode_rewards
```

```python
Q_before = Q.copy()
eval_episode_length, eval_total_reward = evaluation(Q)
print('evaluation episode_length:')
print(eval_episode_length)
print('evaluation total rewards:')
print(eval_total_reward)
print("Unique evaluation lengths:", np.unique(eval_episode_length))
print("Mean evaluation length:", np.mean(eval_episode_length))
```
