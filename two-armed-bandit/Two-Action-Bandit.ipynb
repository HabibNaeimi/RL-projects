```python
import random
import copy
```

# Envirenment

```python
reward_probabilities = [0.2, 0.8]  # For 0, and 1, respectively
```

```python
def get_reward(action):
    """
    Return either 0 or 1 according to the hidden reward probability of the chosen action.
    
    """
    probability = reward_probabilities[action]

    if random.random() < probability:
        return 1
    else:
        return 0
```

# Agent

```python
q_values = [0.0 , 0.0]
epsilon = 0.1
alpha = 0.1
```

```python
def greedy_action(q_values):
    """
    Return the action with the larger estimated Q-value.
    
    """
    if q_values[0] < q_values[1]:
        return int(1)
    elif q_values[0] > q_values[1]:
        return int(0)
    else:
        return int(random.randint(0, 1))
```

```python
def choose_action(q_values, epsilon):
    """
    Choose an action using epsilon-greedy exploration.
    """
    if random.random() < epsilon:
        return int(random.randint(0, 1))
    else:
        return greedy_action(q_values)
```

```python
def update_q(
        q_values,
        action,
        reward,
        alpha,
):
    """
    Update only the Q-value of the action that was taken.
    """
    new_q = copy.deepcopy(q_values)
    prediction_error = reward - q_values[action]
    new_q[action] = q_values[action] + alpha * prediction_error
    return new_q
```

# Training Phase

```python
num_steps = 1000
action_counts = [0, 0]
total_reward = 0
print('Learning curve:')
for step in range(num_steps):
    action = choose_action(q_values=q_values, epsilon=epsilon)
    reward = get_reward(action=action)
    q_values = update_q(
        q_values=q_values,
        action=action,
        reward=reward,
        alpha=alpha
    )
    action_counts[0 if action == 0 else 1] += 1
    total_reward += reward
    if step % 100 == 0:                             # Inspect learning during training
        print(step, q_values, action_counts)     
```

# Results

```python
print("Learned Q-values:", q_values)
print("Action counts:", action_counts)
print("Total reward:", total_reward)
print("Greedy action:", greedy_action(q_values))
```
