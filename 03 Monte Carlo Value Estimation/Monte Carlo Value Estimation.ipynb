```python
import random
import numpy as np
```

# Envirenment

```python
NUM_STATES = 5                           # Number of states
TERMINAL_STATE = 4                          
GAMMA = 0.9

START_STATE = 0
GOAL_STATE = 4

V = np.zeros(NUM_STATES, dtype=float)
visist_counts = np.zeros(NUM_STATES, dtype=int)

LEFT = int(0)
RIGHT = int(1)
```

```python
def step(state, action):
    """
    Applies an action and returns: next_state, reward, terminated
    """
    if action == 0:
        state -= 1
        if state < 0 : state = 0
    elif action == 1:
        state += 1
        if state > GOAL_STATE: state = GOAL_STATE
    
    if state == GOAL_STATE:
        reward = 1
    else:
        reward = 0
    
    if state == GOAL_STATE: terminated = True
    else: terminated = False
    
    return state, reward, terminated
```

# Policy

```python
def choose_action():
    """
    Delivers next action based on Random policy: choose LEFT or RIGHT with equal probability of 50%.
    """
    return random.randint(0, 1)
```

# Episode Generation

```python
def run_episode(max_steps=100):
    """
    Runs one episode and return its trajectory.
    Each trajectory element should be: (state, action, reward)
    """
    state = START_STATE
    trajectory = []
    for _ in range(max_steps):
        action = choose_action()
        next_state, reward, termination = step(state=state, action=action)
        trajectory.append((state, action, reward))
        state = next_state
        if termination: break
    
    return trajectory
```

# Return calculation

```python
def compute_returns(trajectory, gamma=GAMMA):
    """
    Compute discounted return G_t for every transition in the trajectory.
    """
    reversed_returns = []
    for i, val in reversed(list(enumerate(trajectory))):
        reward = val[2]
        if reversed_returns:
            G = reward + gamma * reversed_returns[-1]
            reversed_returns.append(G)
        else:
            reversed_returns.append(reward)
    return list(reversed(reversed_returns))
```

# Update V(s)

```python
def update_values_from_episode(trajectory, V, visit_counts, gamma=GAMMA):
    """
    Update V(s) using every visit in one completed episode.
    """
    returns = compute_returns(trajectory=trajectory, gamma=gamma)

    for trj, ret in zip(trajectory, returns):
        state = trj[0]
        G_t = ret
        visit_counts[state] += 1
        V[state] += (G_t - V[state]) / visit_counts[state]
    
```

# Training

```python
def train_mc(num_spisodes, gamma=GAMMA):
    V = np.zeros(NUM_STATES, dtype=float)
    visit_counts = np.zeros(NUM_STATES, dtype=int)
    
    episode_reterns = []

    for episode in range(num_spisodes):
        trajectory = run_episode()
        update_values_from_episode(trajectory, V, visit_counts, gamma=gamma)
        episode_reterns.append(sum([i[2] for i in trajectory]))
    
    return V, visit_counts, episode_reterns
```

# Runing

```python
ep = [1, 100, 1000, 10000]
res = {}
for e in ep:
    V, visist_counts, episode_returns = train_mc(e)
    res[e] = {"V": V, "visist_counts": visist_counts, "episode_returns": episode_returns}

```
