```python
import random
```

# Envirenment

```python
START_STATE = 0
GOAL_STATE = 4

LEFT = int(0)
RIGHT = int(1)
```

```python
def step(state, action):
    """
    Applies an action and returns: next_state, reward, terminated
    """
    if action == 0:
        state -= 1
        if state < 0 : state = 0
    elif action == 1:
        state += 1
        if state > GOAL_STATE: state = GOAL_STATE
    
    if state == GOAL_STATE:
        reward = 1
    else:
        reward = 0
    
    if state == GOAL_STATE: terminated = True
    else: terminated = False
    
    return state, reward, terminated
```

# Policy

```python
def choose_action():
    """
    Delivers next action based on Random policy: choose LEFT or RIGHT with equal probability of 50%.
    """
    return random.randint(0, 1)
```

# Episode Generation

```python
def run_episode(max_steps=100):
    """
    Runs one episode and return its trajectory.
    Each trajectory element should be: (state, action, reward)
    """
    state = START_STATE
    trajectory = []
    for _ in range(max_steps):
        action = choose_action()
        next_state, reward, termination = step(state=state, action=action)
        trajectory.append((state, action, reward))
        state = next_state
        if termination: break
    
    return trajectory
```

# Return calculation

```python
def compute_returns(trajectory, gamma=0.9):
    """
    Compute discounted return G_t for every transition in the trajectory.
    """
    reversed_returns = []
    for i, val in reversed(list(enumerate(trajectory))):
        reward = val[2]
        if reversed_returns:
            G = reward + gamma * reversed_returns[-1]
            reversed_returns.append(G)
        else:
            reversed_returns.append(reward)
    return list(reversed(reversed_returns))
```

# Runing

```python
trajectory = run_episode()
print("Trajectories:")

for transition in trajectory:
    print(transition)

returns = compute_returns(trajectory=trajectory, gamma=0.9)
print('returns:')
print(returns)
```

# Testing 

```python
assert step(0, LEFT) == (0, 0, False)

assert step(0, RIGHT) == (1, 0, False)

assert step(3, RIGHT) == (4, 1, True)

assert step(2, LEFT) == (1, 0, False)
```

```python
# Environment boundaries
print(step(0, LEFT))
print(step(3, RIGHT))
```

```python
# trajectory alignment Inspection
for transition, G in zip(trajectory, returns):
    print(transition, "-> return:", G)
```
